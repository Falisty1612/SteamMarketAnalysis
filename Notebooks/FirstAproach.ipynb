```python
#Sprawdzenie NULLi tylko w kolumnach numerycznych i datach
numeric_columns = ["appid", "negative_ratings", "average_playtime", "median_playtime", "price", "release_date"]

count_before = stdf.count()
df_dropna = stdf.dropna(subset=numeric_columns)
count_after = df_dropna.count()

if count_before != count_after:
    print(f"Rows with NULLs in numeric columns")
    display(df_dropna)
else:
    print("No NULLs in numeric columns")
```

```python
import pyspark.sql.functions as F
#Nowa kolumna na cene
stdf = stdf.withColumn("price_num", F.col("price").cast("double"))
```

```python
#Konwersja daty na prawidłowy format
stdf = stdf.withColumn("release_da", F.to_date(F.col("release_date"), "yyyy-MM-dd"))
```

```python
stdf.printSchema()
```

```python
#Usunięcie starych kolumn przed transformacją i zastąpienie ich nowymi
stdf = stdf.drop("price")
stdf = stdf.drop("release_date")
stdf = stdf.withColumnRenamed("price_num", "price")
stdf = stdf.withColumnRenamed("release_da", "release_date")
```

```python
spark.catalog.dropTempView("steam")
stdf.createOrReplaceTempView("steam")
spark.sql("SHOW VIEWS").show() #lista wszystkich widoków
```
